```python
import subprocess
import time

# Start Ollama server in background logging to ollama.log
log_file = open("ollama.log", "w")
subprocess.Popen(["ollama", "serve"], stdout=log_file, stderr=subprocess.STDOUT)
time.sleep(10)
print("Ollama server started.")
```
